```python
#projet 1
import random

while True :
 print("Bienvenue dans le jeu !\nDevinez le nombre auquel je pense (entre 1 et 100).")
 n = random.randint(1,5)
 while True :
  x = int(input("Votre réponse : "))
  if x < n:
    print("C'est plus. Essayez encore !")
  elif x > n:
    print("C'est moins")
  else:
    print(f"Bravo ! \nJe pensais en effet à {x} !")
    a = input("Pour rejouer, tapez y")
    if a.lower() == "y":
     break

```
